## Exercise 1 — Introduction to Data Analysis
> **Objective:** Understand the fundamental principles and importance of data analysis.


### What is Data Analysis?

Data analysis is the systematic process of **collecting, cleaning, transforming, and modeling data** 
with the goal of extracting useful information, drawing conclusions, and supporting decision-making. 
It spans a wide range of techniques — from simple descriptive statistics to complex predictive modeling — 
and applies to both quantitative and qualitative data.

### Why is Data Analysis Important in Modern Contexts?

In today's data-driven world, organizations generate vast amounts of information every second. 
Without the ability to analyze and interpret this data, it remains raw and unexploitable. 
Data analysis enables businesses and researchers to:

- **Identify patterns and trends** that are invisible to the naked eye
- **Reduce uncertainty** in decision-making by replacing intuition with evidence
- **Optimize operations** by detecting inefficiencies and areas for improvement
- **Predict future outcomes** using historical data and statistical models
- **Personalize experiences** for customers, users, or patients

### Three Real-World Application Domains

**1. Healthcare & Medicine**  
Medical researchers and hospitals use data analysis to predict patient outcomes, identify risk factors for diseases, 
and optimize treatment protocols. Analyzing electronic health records, for instance, helps detect early signs of sepsis 
and significantly reduce mortality rates.

**2. E-Commerce & Retail**  
Platforms like Amazon analyze browsing history, purchase behavior, and customer reviews to build recommendation engines, 
personalize the shopping experience, and forecast inventory needs — directly increasing revenue and customer retention.

**3. Finance & Banking**  
Financial institutions analyze transaction data in real time to detect fraudulent activity, assess credit risk, 
and automate trading strategies. Credit scoring models evaluate hundreds of variables to predict the likelihood 
that a borrower will default on a loan.

## Exercise 4 — Exploring Data Types (Iris Dataset)
> **Objective:** Identify qualitative and quantitative columns in the Iris dataset.


```python
iris_df = pd.read_csv('datasets/Iris.csv')

print('Shape:', iris_df.shape)
print('\n--- First rows ---')
display(iris_df.head())
print('\n--- Data types ---')
print(iris_df.dtypes)
print('\n--- Unique values per column ---')
for col in iris_df.columns:
    print(f'  {col}: {iris_df[col].nunique()} unique value(s)')
```

### Column Classification

| Column | dtype | Classification | Reasoning |
|---|---|---|---|
| Id | int64 | Quantitative (discrete) | Row identifier — integer with no real analytical meaning |
| SepalLengthCm | float64 | **Quantitative (continuous)** | Physical measurement in cm, infinite possible values |
| SepalWidthCm | float64 | **Quantitative (continuous)** | Physical measurement in cm |
| PetalLengthCm | float64 | **Quantitative (continuous)** | Physical measurement in cm |
| PetalWidthCm | float64 | **Quantitative (continuous)** | Physical measurement in cm |
| Species | object | **Qualitative (nominal)** | Named class (Iris-setosa, Iris-versicolor, Iris-virginica) — no natural order |


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

palette = {'Iris-setosa': '#E50914', 'Iris-versicolor': '#F5A623', 'Iris-virginica': '#4A90D9'}
for species, grp in iris_df.groupby('Species'):
    axes[0].scatter(grp['SepalLengthCm'], grp['PetalLengthCm'],
                    label=species, color=palette[species], alpha=0.8, s=50)
axes[0].set_xlabel('Sepal Length (cm)')
axes[0].set_ylabel('Petal Length (cm)')
axes[0].set_title('Quantitative: Sepal vs Petal Length by Species')
axes[0].legend()

counts = iris_df['Species'].value_counts()
axes[1].bar(counts.index, counts.values,
            color=[palette[s] for s in counts.index], edgecolor='white')
axes[1].set_title('Qualitative (nominal): Species Distribution')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

## Exercise 5 — Basic Data Observation and Analysis Skills
> **Objective:** Develop observation skills using the Sleep dataset.


```python
sleep_df = pd.read_csv('datasets/Time Americans Spend Sleeping.csv')

print('Shape:', sleep_df.shape)
print('\nNull values:')
print(sleep_df.isnull().sum())
print('\nColumn types:')
print(sleep_df.dtypes)
print('\nUnique values per column:')
for col in sleep_df.columns:
    print(f'  {col}: {sleep_df[col].unique()[:5]}')
```

### Relevant Columns for Different Analysis Types

**1. Trend Analysis** — *How has average sleep duration evolved over years?*
- Relevant columns: `Year`, `Avg hrs per day sleeping`
- Justification: `Year` is the time dimension; `Avg hrs per day sleeping` is the measured outcome. 
A line chart over years reveals whether Americans sleep more or less over time.

**2. Group Comparison** — *Do men and women sleep differently?*
- Relevant columns: `Sex`, `Avg hrs per day sleeping`, `Age Group`
- Justification: `Sex` (nominal) acts as the grouping variable; sleep duration is compared across groups 
using bar charts or box plots.

**3. Weekday vs Weekend Comparison** — *Do Americans sleep more on weekends?*
- Relevant columns: `Type of Days`, `Avg hrs per day sleeping`
- Justification: `Type of Days` separates weekday from weekend observations, 
allowing a direct comparison of average sleep duration across day types.


```python
# Filter to clean comparable subsets
trend = sleep_df[(sleep_df['Sex'] == 'Both') & 
                 (sleep_df['Age Group'] == '15 years and over') &
                 (sleep_df['Type of Days'] == 'All days')].copy()

gender = sleep_df[(sleep_df['Type of Days'] == 'All days') &
                  (sleep_df['Age Group'] == '15 years and over') &
                  (sleep_df['Sex'].isin(['Men', 'Women']))].copy()

daytype = sleep_df[(sleep_df['Sex'] == 'Both') &
                   (sleep_df['Age Group'] == '15 years and over') &
                   (sleep_df['Type of Days'].isin(['Weekdays', 'Weekends']))].copy()

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

# Plot 1: Trend over years
axes[0].plot(trend['Year'], trend['Avg hrs per day sleeping'], marker='o', color='#4A90D9')
axes[0].set_title('Trend: Avg Sleep Duration Over Years')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Avg hrs/day')

# Plot 2: Gender comparison
gender_avg = gender.groupby('Sex')['Avg hrs per day sleeping'].mean()
axes[1].bar(gender_avg.index, gender_avg.values, color=['#E50914','#F5A623'], edgecolor='white')
axes[1].set_title('Group Comparison: Sleep by Gender')
axes[1].set_ylabel('Avg hrs/day')
axes[1].set_ylim(8, 9)

# Plot 3: Weekday vs Weekend
dt_avg = daytype.groupby('Type of Days')['Avg hrs per day sleeping'].mean()
axes[2].bar(dt_avg.index, dt_avg.values, color=['#2ECC71','#9B59B6'], edgecolor='white')
axes[2].set_title('Weekdays vs Weekends Sleep Duration')
axes[2].set_ylabel('Avg hrs/day')
axes[2].set_ylim(8, 9.5)

plt.tight_layout()
plt.show()
```

## Exercise 7 — Unstructured to Structured Data Transformation
> **Objective:** Propose a conversion method for each unstructured data source.


### 1. Blog articles about travel experiences
**Method:** Natural Language Processing (NLP) + Named Entity Recognition (NER)  
Extract: destination names, dates, sentiment scores, activities mentioned.  
**Output table:** `[article_id, destination, travel_date, sentiment_score, activities]`  
**Justification:** NER reliably extracts location and date entities; sentiment analysis converts 
subjective prose into a numeric score, making the content comparable across articles.

### 2. Audio recordings of customer service calls
**Method:** Automatic Speech Recognition (ASR) → Text → NLP pipeline  
Step 1: Transcribe audio to text (e.g. Whisper, Google Speech-to-Text).  
Step 2: Extract structured fields — duration, caller intent, keywords, sentiment, resolution.  
**Output table:** `[call_id, duration_sec, intent, sentiment, resolved, keywords]`  
**Justification:** ASR bridges the gap between audio and text; NLP then converts free text into structured fields.

### 3. Handwritten notes from a brainstorming session
**Method:** Optical Character Recognition (OCR) → Text classification  
Step 1: Scan/photograph notes and run OCR (e.g. Tesseract, Google Vision API).  
Step 2: Classify extracted items by topic, priority, or type.  
**Output table:** `[note_id, text, category, priority_level, author]`  
**Justification:** OCR converts visual text into machine-readable strings; classification adds the structure.

### 4. A cooking tutorial video
**Method:** Computer Vision + Audio Transcription + Temporal segmentation  
Extract frames at key intervals (object/action recognition), transcribe audio narration, 
segment video into steps based on scene changes.  
**Output table:** `[step_number, start_time, end_time, action, ingredients_shown, instruction_text]`  
**Justification:** Combining vision and audio modalities yields the richest structured output, 
capturing both what is shown and what is said at each step.

## Exercise 8 — Importing the Titanic Dataset from GitHub
> **Objective:** Load the Titanic training dataset and display the first rows.


```python
# Dataset originally from Kaggle, hosted on GitHub for easy access.
# To load directly from GitHub URL:
# url = 'https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/...train.zip'
# train_df = pd.read_csv(url)

train_df = pd.read_csv('datasets/train.csv')

print('Shape:', train_df.shape)
print('\n--- First rows ---')
display(train_df.head())
print('\n--- Data types ---')
print(train_df.dtypes)
print('\n--- Missing values ---')
print(train_df.isnull().sum())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Survival
sv = train_df['Survived'].value_counts()
axes[0].bar(['Not Survived', 'Survived'], sv.values, color=['#E50914','#2ECC71'], edgecolor='white')
axes[0].set_title('Survival Distribution')
axes[0].set_ylabel('Count')

# Passenger class
pc = train_df['Pclass'].value_counts().sort_index()
axes[1].bar([f'Class {c}' for c in pc.index], pc.values,
            color=['#4A90D9','#F5A623','#E50914'], edgecolor='white')
axes[1].set_title('Passenger Class Distribution')
axes[1].set_ylabel('Count')

# Age distribution
train_df['Age'].dropna().plot(kind='hist', bins=20, ax=axes[2], color='#9B59B6', edgecolor='white')
axes[2].set_title('Age Distribution')
axes[2].set_xlabel('Age')
axes[2].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
